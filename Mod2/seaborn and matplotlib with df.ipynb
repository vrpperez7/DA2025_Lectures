```python
import matplotlib.pyplot as plt
 
weeks = ['Week 1', 'Week 2', 'Week 3', 'Week 4', 'Week 5']
downloads = [230, 310, 390, 500, 450] # Sample data with a dip in Week 5
 
plt.plot(weeks, downloads, marker='o', linestyle='-', color='skyblue')
plt.title('Weekly App Downloads Over Time')
plt.xlabel('Week')
plt.ylabel('Downloads')
plt.grid(True, linestyle='--', alpha=0.7) # Add a subtle grid
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample Data for the Try-It-Out activity
days = list(range(1, 31))
visitors = [500, 520, 550, 510, 600, 620, 580, 590, 630, 650, # Week 1-2
            400, 410, 430, 390, 380, 370, 350, 360, 340, 330, # Week 3-4 (simulated dip)
            500, 510, 530, 550, 570, 590, 610, 630, 650, 670] # Week 5-6 (recovery)
signups = [50, 52, 55, 51, 60, 62, 58, 59, 63, 65,
           30, 31, 33, 29, 28, 27, 25, 26, 24, 23,
           45, 46, 48, 50, 52, 54, 56, 58, 60, 62]

# Activity 1: Visitors Trend (Line Plot)
plt.figure(figsize=(10, 6))
plt.plot(days, visitors, marker='o', linestyle='-', color='darkblue')
plt.title('Daily Visitors Over Time')
plt.xlabel('Day of Month')
plt.ylabel('Number of Visitors')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(range(1, 31, 2)) # Show every other day for clarity
plt.tight_layout()
plt.show()

```

```python
# Activity 2: Sign-up Distribution (Histogram)
plt.figure(figsize=(8, 5))
plt.hist(signups, bins=None, edgecolor='black', alpha=0.7, color='teal')
plt.title('Distribution of Daily Sign-ups')
plt.xlabel('Number of Sign-ups')
plt.ylabel('None')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```

```python
# Activity 3: Relationship (Scatter Plot)
plt.figure(figsize=(7, 5))
plt.scatter(visitors, signups, color='purple', alpha=0.7)
plt.title('Website Visitors vs. New Sign-ups')
plt.xlabel('Number of Visitors')
plt.ylabel('Number of Sign-ups')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample data
data = pd.DataFrame({
    'Department': ['Sales', 'Engineering', 'HR', 'Marketing', 'Sales', 'Engineering', 'HR', 'Marketing'],
    'Satisfaction': [3.9, 4.2, 4.0, 3.6, 4.1, 4.3, 3.8, 3.7],
    'Employee_Count': [50, 75, 20, 30, 55, 70, 25, 35] # Added for potential second plot
})

plt.figure(figsize=(8, 6))
sns.barplot(data=data, x='Department', y='Satisfaction', palette='viridis', ci=None)
plt.title('Average Satisfaction Score by Department')
plt.xlabel('Department')
plt.ylabel('Average Satisfaction Score (1-5)')
plt.ylim(3, 4.5) # Set y-axis limits for better focus
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Sample Data for the Try-It-Out activity
# Simulating ratings for three features with different distributions
np.random.seed(42)
data_ratings = {
    'Feature': (['Search'] * 100) + (['Checkout'] * 100) + (['Recommendations'] * 100),
    'Rating': np.concatenate([
        np.random.randint(3, 6, 100),  # Search: mostly 3, 4, 5
        np.random.choice([1, 2, 4, 5], 100, p=[0.2, 0.2, 0.3, 0.3]), # Checkout: bimodal
        np.random.randint(1, 4, 100)  # Recommendations: mostly 1, 2, 3
    ])
}
df_ratings = pd.DataFrame(data_ratings)

# Activity: Bar Plot for Average Rating
plt.figure(figsize=(8, 5))
sns.barplot(data=df_ratings, x='Feature', y='Rating', estimator=np.mean, palette='deep', ci=None)
plt.title('Average Customer Rating by Feature')
plt.xlabel('Feature')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show() 
```
