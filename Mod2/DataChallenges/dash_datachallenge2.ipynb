```python
import pandas as pd
import dash
from dash import html, dcc
from dash.dependencies import Input, Output
import plotly.express as px

```

```python
df = pd.read_csv("/Users/beans/Desktop/Marcy_Lab/DA2025_Lectures/Mod2/data/indian_food.csv").dropna()
```

```python
#grouping course by both mean of prep time and cook time
courseprep = df.groupby('course').agg({'prep_time':['mean'], 'cook_time':['mean']})
#courseprep.head()
#created multi index columns so to fix we renamed columns and reset the index
courseprep.columns = ['prep_time','cook_time']
courseprep = courseprep.reset_index()
courseprep.columns
```

```python
#just testing
fig = px.bar(courseprep, x='course', y='prep_time')
fig.show()
```

```python
app = dash.Dash(__name__)
app.title = "Name"

app.layout = html.Div([
    html.H1("Indian Food with Dropdown"),
dcc.Dropdown(
    id='preporcook',
    options=[{'label': 'prep_time', 'value':'prep_time'}, {'label': 'cook_time', 'value':'cook_time'}],
    placeholder = 'select an option',
    style={'width': '50%'}
    ),
    dcc.Graph(id='bar')
])
```

```python
@app.callback(
    Output('bar', 'figure'),
    Input('preporcook', 'value')
)
#used chart selected value to display bar chart based on input
def update_chart(selected_value):
    fig = px.bar(courseprep, x='course', y = selected_value)
    return fig

if __name__ == "__main__":
    app.run(debug=True)
```
